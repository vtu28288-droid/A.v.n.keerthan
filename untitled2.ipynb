```python
from collections import deque
def bfs(graph, start):
    visited = set();
    queue = deque([start])
    visited.add(start)
    while queue:
        node = queue.popleft()
        print(node, end=" ")
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
graph = {
    'A':['B','C'],
    'B':[],
    'C':[]
}
print("BFS Traversal:")
bfs(graph, 'A')
```

```python


heights = [1, 3, 5, 7, 6, 8, 9, 7, 5, 10]

current = 0

print("Starting position:", current)
print("Starting height:", heights[current])

while True:
    left = current - 1
    right = current + 1
    best = current
    if left >= 0 and heights[left] > heights[best]:
        best = left
    if right < len(heights) and heights[right] > heights[best]:
        best = right
    if best == current:
        print("\nNo better neighbor found!")
        print("Hill climbing stopped.")
        break
    current = best

    print("Moved to position:", current,
          "Height:", heights[current])

print("\nFinal position:", current)
print("Final height:", heights[current])
```

```python
def find_all_paths(rows, cols, start, end):
    directions = [
        (1, 0),
        (-1, 0),
        (0, 1),
        (0, -1)
    ]

    all_paths = []

    def dfs(current, path, visited):
        if current == end:
            all_paths.append(path.copy())
            return

        r, c = current

        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            next_cell = (nr, nc)

            if (0 <= nr < rows and
                0 <= nc < cols and
                next_cell not in visited):

                visited.add(next_cell)
                path.append(next_cell)

                dfs(next_cell, path, visited)


                path.pop()
                visited.remove(next_cell)

    dfs(start, [start], {start})

    return all_paths


paths = find_all_paths(3, 3, (0, 0), (1,2)
for i, path in enumerate(paths, 1):
    print(f"Path {i}: {path}")
    print(f"Path length: {len(path) - 1}")
    print()
```

```python
import heapq

def heuristic(a, b):

    return abs(a[0] - b[0]) + abs(a[1] - b[1])


def a_star(rows, cols, start, goal):

    open_list = []
    heapq.heappush(open_list, (0, 0, start, [start]))

    g_cost = {start: 0}

    directions = [
        (0, 1),
        (1, 0),
        (0, -1),
        (-1, 0)
    ]

    while open_list:
        f, g, current, path = heapq.heappop(open_list)

        if current == goal:
            return path, g

        for dr, dc in directions:
            nr = current[0] + dr
            nc = current[1] + dc
            neighbor = (nr, nc)

            if 0 <= nr < rows and 0 <= nc < cols:

                new_g = g + 1


                if neighbor not in g_cost or new_g < g_cost[neighbor]:
                    g_cost[neighbor] = new_g

                    h = heuristic(neighbor, goal)
                    f = new_g + h

                    heapq.heappush(
                        open_list,
                        (f, new_g, neighbor, path + [neighbor])
                    )

    return None, None



rows = int(input("enter number of row:"))
cols = int(input("enter number of coloumn:"))

start = (0, 0)
goal = (rows - 1, cols - 1)

print("start", start)
print("goal =", goal)


path, cost = a_star(rows, cols, start, goal)

if path:
    print("\noptimal path:", end="")
    print(path)
    print("cost :", cost)
else:
    print("No path found")
```

```python
def minimax(values, depth, index, maximizing_player):
    if depth == 0:
        return values[index]
    if maximizing_player:
        return max(
            minimax(values, depth - 1, index * 2, False),
            minimax(values, depth - 1, index * 2 + 1, False)
        )
    else:
        return min(
            minimax(values, depth - 1, index * 2, True),
            minimax(values, depth - 1, index * 2 + 1, True)
        )
height = int(input("enter tree height:"))
n = 2 ** height
print("enter", n, "leaf node values:")
values = list(map(int, input().split()))
result = minimax(values, height, 0, True)
print("leaf values :", values)
print("min max value :", result)
```
